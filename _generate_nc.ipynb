```python
import calliope
import os

BASE = os.path.dirname(os.path.abspath("_generate_nc.ipynb"))
EXAMPLES = {
    "national_scale": os.path.join(BASE, "examples", "national_scale", "model.yaml"),
    "urban_scale": os.path.join(BASE, "examples", "urban_scale", "model.yaml"),
}

for name, model_path in EXAMPLES.items():
    print(f"\n=== {name} ===")
    model = calliope.read_yaml(model_path)
    model.build()
    model.solve()
    term = model.results.attrs.get("termination_condition", "unknown")
    print(f"Termination: {term}")
    assert term in ("optimal", "feasible"), f"Solver did not exit normally: {term}"
    out = os.path.join(BASE, "examples", name, f"{name}_results.nc")
    model.to_netcdf(out)
    print(f"Saved: {out}")

print("\nDone.")
```
